# 1. Figure per il capitolo di tesi
Genera 13 PDF vettoriali e 13 PNG dai risultati congelati in `source_data/`. Non esegue training, inferenza o nuovi confronti sul test. Tutte le etichette nelle figure sono in inglese.
La cartella `thesis_figures` puo' essere copiata da sola nella macchina di produzione. Dipendenze: NumPy e Matplotlib.

```python
from pathlib import Path
import sys
import subprocess

THESIS_FIGURES = None  
if THESIS_FIGURES is None:
    candidates = [Path.cwd(), Path.cwd() / 'thesis_figures',
                  Path('/content/ML_metrologia/thesis_figures'),
                  Path('/content/thesis_figures')]
    candidates += [p / 'thesis_figures' for p in Path.cwd().parents]
    THESIS_FIGURES = next((p for p in candidates if (p / 'make_thesis_figures.py').is_file()), None)
if THESIS_FIGURES is None:
    raise FileNotFoundError('Set THESIS_FIGURES to the copied figure directory.')
THESIS_FIGURES = Path(THESIS_FIGURES).resolve()
print(THESIS_FIGURES)
```

```python
subprocess.run([sys.executable, str(THESIS_FIGURES / 'make_thesis_figures.py')], check=True)
```

```python
import json
from IPython.display import display, Image
manifest = json.loads((THESIS_FIGURES / 'generation_manifest.json').read_text())
for stem in manifest['figures']:
    print(stem)
    display(Image(filename=str(THESIS_FIGURES / f'{stem}.png'), width=850))
```
